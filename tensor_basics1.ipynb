## metrices as tensor

```python
q1 = torch.tensor(
    [
        [200, 220, 250],
        [150, 180, 210],
        [300, 330, 360]
    ]
    
)


q2 = torch.tensor([
    [209, 231, 259],
    [155, 192, 222],
    [310, 340, 275]
])
```

```python
## total revenue from q1 and q2

revenue = q1 + q2
revenue
```

```python
## revenue growth in q2 as compared to q1

revenue_growth2 = q2 - q1
revenue_growth2
```

```python
## sales unit growth in q2 as compared to q1

salesunit_growth2 = ((q2-q1)/q1)*100
salesunit_growth2
```
