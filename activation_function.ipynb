```python
import pandas as pd
import numpy as np
```

```python
def sigmoid(x):
    return 1/(1 + np.exp(-x))
```

```python
sigmoid(2)
```

```python
values=np.array([-2,-1,0, 1, 2, 10, 50, 500])
sigmoid(values)
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / exp_x.sum(axis=0)

values = np.array([3, 8, 9, 7])
softmax(values)
```

```python
softmax(values).sum()
```
